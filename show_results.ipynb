```python
import os
import torch

DIR = 'riperino'
KEY = ['test_voterproba_metrics']
BOLD = "\033[1m"
RED = "\033[31m"
GREEN = "\033[32m"
MAGENTA = "\033[35m"
CYAN = "\033[36m"
RESET = "\033[0m"


def extract_scores(pytorch_model, run_metric):
    try:
        return pytorch_model[run_metric]["F1_SCORE_MACRO"]
    except KeyError:
        return 0
        
def calculate_avg(values):
    return 0 if any(v == 0 for v in values) else sum(values) / len(values)


path = os.path.join(os.getcwd(), DIR)
score_dict = {}
valid_dict = {}
for subfolder in os.listdir(DIR):
    case_path = os.path.join(path, subfolder)  
    names = [os.path.join(case_path, name) for name in os.listdir(case_path)]
    models = [torch.load(name) for name in names]
    f1_score = list(map(extract_scores, models, [KEY[-1]] * len(models)))
    score_dict[os.path.basename(case_path)] = f1_score
        
    
average_dict = dict(zip(score_dict.keys(), map(calculate_avg, score_dict.values())))

for key, value in average_dict.items():
    print(f'{BOLD}{RED}{RESET}{key}: {MAGENTA}{value:.3f}{RESET}')
```
